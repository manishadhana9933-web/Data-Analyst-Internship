```python
total_units = df["Units Sold"].sum()

print("Total Units Sold:", total_units)
```

```python
total_gross_sales = df["Gross Sales"].sum()

print("Total Gross Sales:", total_gross_sales)
```

```python
total_discounts = df["Discounts"].sum()

print("Total Discounts:", total_discounts)
```

```python
total_sales = df["Sales"].sum()

print("Total Sales:", total_sales)
```

```python
total_cogs = df["COGS"].sum()

print("Total COGS:", total_cogs)
```

```python
total_profit = df["Profit"].sum()

print("Total Profit:", total_profit)
```

```python
profit_margin = (total_profit / total_sales) * 100

print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
average_sale_price = df["Sale Price"].mean()

print("Average Sale Price:", round(average_sale_price, 2))
```

```python
segment_sales = (
    df.groupby("Segment")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

print(segment_sales)
```

```python
segment_profit = (
    df.groupby("Segment")["Profit"]
      .sum()
      .sort_values(ascending=False)
)

print(segment_profit)
```
